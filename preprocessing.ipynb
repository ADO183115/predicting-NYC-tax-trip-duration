```python
import pandas as pd
```

```python
train_df = pd.read_pickle("train.pkl")
test_df = pd.read_pickle("test.pkl")
```

```python
train_df.shape
```

### Train Data

```python
# Prune rows where trip duration is less than 60 or higher than 7200.

train_df = train_df.query("(60 < trip_duration < 7200)")

train_df.shape[0]
```

```python
# Prune rows where longitude is not between -73 and -74.5.
# Prune rows where latitude is not between 40 and 41.

train_df = train_df.query("(-74.5 < pickup_longitude < -73)")
train_df = train_df.query("(-74.5 < dropoff_longitude < -73)")

train_df = train_df.query("(40 < pickup_latitude < 41)")
train_df = train_df.query("(40 < dropoff_latitude < 41)")

train_df.shape[0]
```

```python
# Prune rows where trip distance is more than 60.

train_df = train_df.query("trip_distance < 60")

train_df.shape[0]
```

```python
# Add "is_rush_hour" column.

train_df["is_rush_hour"] = (((14 <= train_df["pickup_hour"]) & (train_df["pickup_hour"] <= 16)) \
                            & ((0 <= train_df["pickup_day_of_week"]) & (train_df["pickup_day_of_week"] <= 5))).astype(bool)

train_df[train_df["is_rush_hour"] == True].shape[0]

train_df.head()
```

```python
train_df = train_df.drop(columns=["id", "pickup_datetime", "dropoff_datetime", "store_and_fwd_flag"])

train_df.head()
```

### Test Data

```python
# Add "is_rush_hour" column.

test_df["is_rush_hour"] = (((14 <= test_df["pickup_hour"]) & (test_df["pickup_hour"] <= 16)) \
                            & ((0 <= test_df["pickup_day_of_week"]) & (test_df["pickup_day_of_week"] <= 5))).astype(bool)

test_df[test_df["is_rush_hour"] == True].shape[0]

test_df.head()
```

```python
# Save test.csv "id" column seperately for submission.

test_df_id_col = test_df["id"]

test_df_id_col
```

```python
test_df = test_df.drop(columns=["id", "pickup_datetime", "store_and_fwd_flag"])

test_df.head()
```
